### Title: Demonstration of the Law of Large Numbers
### Objectives
* To understand the concept of the Law of Large Numbers.
* To simulate random number generation between 1 and 10 using Python.
* To observe how the running average converges to the theoretical expected value as the number of trials increases.

### Theory

The Law of Large Numbers (LLN) is one of the fundamental theorems of probability and statistics. It states that as the number of independent and identical random experiments increases, the sample average approaches the expected (true) value of the random variable.

Suppose a random variable X has an expected value E[X]. If the experiment is repeated n times, the sample mean is given by:

Xˉ=(X1​+X2​+⋯+Xn)/n​​
	​
According to the Law of Large Numbers,

n→∞lim​Xˉ=E[X]

In this experiment, random integers between 1 and 10 are generated repeatedly. Since each number has an equal probability of occurring, the theoretical expected value is

E[X]=(1+2+⋯+10)/10=5.5

After each random number is generated, the cumulative (running) average is calculated. Initially, the running average fluctuates significantly because of the small number of observations. As the number of trials increases, these fluctuations decrease, and the running average converges toward the theoretical mean of 5.5, illustrating the Law of Large Numbers.

The convergence is visualized using a graph that plots the running average against the number of trials, with a horizontal reference line representing the theoretical expected value.

```python
"""
Law of Large Numbers Demo
--------------------------
We roll a random number between 1 and 10 (like a 10-sided die), 10,000 times.
After each roll we compute the RUNNING AVERAGE of all rolls so far.

Theory says: as the number of trials -> infinity, the running average
should converge to the true expected value:

    E[X] = (1 + 2 + ... + 10) / 10 = 5.5

The graph shows the running average settling down near 5.5 as the
number of trials grows.
"""

import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Step 1: Generate 10,000 random integers between 1 and 10
# ---------------------------------------------------------
np.random.seed(42)          # remove this line if you want different results each run
N = 10000
rolls = np.random.randint(1, 11, size=N)   # random integers in [1, 10]

# ---------------------------------------------------------
# Step 2: Compute the running (cumulative) average at each step
# ---------------------------------------------------------
cumulative_sum = np.cumsum(rolls)
trial_numbers = np.arange(1, N + 1)
running_average = cumulative_sum / trial_numbers

# ---------------------------------------------------------
# Step 3: Print some summary numbers
# ---------------------------------------------------------
true_mean = (1 + 10) / 2   # = 5.5 for a uniform 1..10 distribution
final_average = running_average[-1]

print(f"Theoretical expected value : {true_mean}")
print(f"Average after 10 trials    : {running_average[9]:.4f}")
print(f"Average after 100 trials   : {running_average[99]:.4f}")
print(f"Average after 1,000 trials : {running_average[999]:.4f}")
print(f"Average after 10,000 trials: {final_average:.4f}")

# ---------------------------------------------------------
# Step 4: Plot the running average vs number of trials
# ---------------------------------------------------------
plt.figure(figsize=(10, 6))
plt.plot(trial_numbers, running_average, color="steelblue", linewidth=1.2,
         label="Running average of rolls")
plt.axhline(y=true_mean, color="red", linestyle="--", linewidth=1.5,
            label=f"True mean = {true_mean}")

plt.xscale("log")   # log scale makes early fluctuations & convergence easier to see
plt.xlabel("Number of trials (log scale)")
plt.ylabel("Running average")
plt.title("Law of Large Numbers: Random numbers 1–10, up to 10,000 trials")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig("law_of_large_numbers.png", dpi=150)
plt.show()
```

### Discussion

In this experiment, 10,000 random integers between 1 and 10 were generated. The running average was calculated after each trial and compared with the theoretical expected value of 5.5.

At the beginning of the simulation, the running average showed noticeable fluctuations due to the small number of observations. However, as more random numbers were generated, the fluctuations gradually decreased, and the running average stabilized near 5.5. This behavior confirms the Law of Large Numbers, which states that the sample mean approaches the expected value as the sample size increases.

The logarithmic scale used on the x-axis clearly illustrates both the large fluctuations during the initial trials and the gradual convergence as the number of trials grows. Minor differences between the final average and the theoretical mean occur because random sampling is subject to natural statistical variation.

### Conclusion

The experiment successfully demonstrated the Law of Large Numbers using Python. Random numbers between 1 and 10 were generated repeatedly, and their running average was computed. The simulation showed that as the number of trials increased, the running average converged toward the theoretical expected value of 5.5. This verifies that increasing the sample size improves the accuracy of the sample mean and reduces the effect of random fluctuations, confirming the validity of the Law of Large Numbers.
